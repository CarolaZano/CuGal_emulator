### Background Cubic Galileon

```python
import numpy as np
import scipy
import pyccl as ccl
import matplotlib.pyplot as plt
```

```python
"""For tests"""

cosmo_testing = ccl.Cosmology(Omega_c = (0.1417493400784 - 0.0223)/0.67**2,
                           Omega_b =  0.0223/0.67**2,
                           h = 0.67,
                           n_s = 0.9665,
                           A_s = 2.01e-9,
                           Omega_g = 0.0)
```

```python
# For today formalism 
def comp_k_1T(Omg_r, Omg_m, f_phi):
    return 6*f_phi*(Omg_m + Omg_r - 1)

def comp_E_LCDM(x, Omega_r0, Omega_m0):
    Omega_L0 = 1.-Omega_m0-Omega_r0
    E = np.sqrt(Omega_m0*np.exp(-3.*x) + Omega_r0*np.exp(-4.*x) + Omega_L0)
    return E
    
def comp_Omega_r(x, E, Omega_r0, Omega_m0, f_phi):
    Omega_r = Omega_r0*np.exp(-4.*x)/E/E
    return Omega_r

def comp_Omega_m(x, E, Omega_r0, Omega_m0, f_phi):
    Omega_m = Omega_m0*np.exp(-3.*x)/E/E
    return Omega_m

def comp_Omega_L(x, E, Omega_r0, Omega_m0, f_phi):
    Omega_phi0 = (1.-Omega_m0-Omega_r0)*f_phi
    Omega_L0 = 1.-Omega_m0-Omega_r0-Omega_phi0
    Omega_L = Omega_L0/E/E
    return Omega_L

def comp_g_31(Omega_r0, Omega_m0, f_phi):
    k_1 = comp_k_1T(Omega_r0, Omega_m0, f_phi)
    Omega_phi0 = (1.- Omega_m0-Omega_r0)*f_phi

    return Omega_phi0 - k_1/6

"""
def comp_Omega_phi(x, E,phi_prime, Omega_r0, Omega_m0, f_phi):
    Omega_phi0 = (1.-Omega_m0-Omega_r0)*f_phi

    Omega_phi = Omega_phi0/E/E
    return Omega_phi
"""
def comp_Omega_phi(x, E, phi_prime, Omega_r0, Omega_m0, f_phi):
    k_1 = comp_k_1T(Omega_r0, Omega_m0, f_phi)
    g_31 = comp_g_31(Omega_r0, Omega_m0, f_phi)

    Omega_phi = 2/3 * (k_1*phi_prime**2/4 + 3* g_31*E**2*phi_prime**3/2)

    return Omega_phi
"""
def comp_Omega_phi(x, E,phi_prime, Omega_r0, Omega_m0, f_phi):
    Omega_phi = 1 - comp_Omega_L(x, E, Omega_r0, Omega_m0, f_phi) - comp_Omega_m(x, E, Omega_r0, Omega_m0, f_phi)\
                - comp_Omega_r(x, E, Omega_r0, Omega_m0, f_phi)

    return Omega_phi

"""

def comp_E_prime(x, E, phi_prime, Omega_r0, Omega_m0, f_phi):

    g_31 = comp_g_31(Omega_r0, Omega_m0, f_phi)
    k_1 = comp_k_1T(Omega_r0, Omega_m0, f_phi)

    Omega_r_val = comp_Omega_r(x, E,Omega_r0, Omega_m0, f_phi)
    Omega_L_val =comp_Omega_L(x, E, Omega_r0, Omega_m0, f_phi)
    
    term1 = (k_1*E*phi_prime*phi_prime/2 - E*(Omega_r_val - 3*Omega_L_val + 3)) * (k_1/phi_prime + 6*g_31*E*E)
    term2 = 9*g_31**2*E**5*phi_prime**3 + 3*g_31*k_1*E**3*phi_prime**2
    denominator = 2*(k_1 + 6*g_31*E*E) - 3*g_31*g_31*E**4*phi_prime**3
    
    """term1 = (k_1*E*phi_prime*phi_prime/2 - E*(Omega_r_val - 3*Omega_L_val + 3)) * (k_1 + 6*g_31*E*E*phi_prime)
    term2 = 9*g_31**2*E**5*phi_prime**4 + 3*g_31*k_1*E**3*phi_prime**3
    denominator = 2*(k_1 + 6*g_31*E*E*phi_prime) - 3*g_31*g_31*E**4*phi_prime**4
    """
    return (term1+term2)/denominator

def comp_phi_primeprime(x, E, phi_prime, Omega_r0, Omega_m0, f_phi):
    g_31 = comp_g_31(Omega_r0, Omega_m0, f_phi)
    k_1 = comp_k_1T(Omega_r0, Omega_m0, f_phi)
    
    E_prime_val = comp_E_prime(x, E, phi_prime, Omega_r0, Omega_m0, f_phi)
    
    nominator = -(3*g_31*E*phi_prime*phi_prime*(E_prime_val + 3*E) + 3*k_1*phi_prime)
    denominator = k_1 + 6*g_31*E*E*phi_prime
    return nominator/denominator - E_prime_val*phi_prime/E


def compute_derivatives(x, Y, Omega_r0, Omega_m0, f_phi):
    phi_primeY, EY = Y
    return [comp_phi_primeprime(x, EY, phi_primeY, Omega_r0, Omega_m0, f_phi), \
            comp_E_prime(x, EY, phi_primeY, Omega_r0, Omega_m0, f_phi)]


z_final = 0.
x_ini = np.log(1./(1.+49))
x_final = np.log(1./(1.+z_final))
x_arr = np.linspace(x_final, x_ini, 500)

Y0 = [1,1]

f_phi = 0.1

ans = scipy.integrate.solve_ivp(compute_derivatives,[x_final,x_ini], Y0, t_eval=x_arr, method='RK45', args=(cosmo_testing["Omega_g"], \
                cosmo_testing["Omega_m"], f_phi), rtol = 2e-14, atol = 2e-14)
```

### Different method - solve for E beforehand

```python
# For today formalism 
def comp_k_1T(Omg_r, Omg_m, f_phi):
    return 6*f_phi*(Omg_m + Omg_r - 1)

def comp_g_31(Omega_r0, Omega_m0, f_phi):
    k_1 = comp_k_1T(Omega_r0, Omega_m0, f_phi)
    Omega_phi0 = (1.- Omega_m0-Omega_r0)*f_phi

    return Omega_phi0 - k_1/6

def comp_E_LCDM(x, Omega_r0, Omega_m0):
    Omega_L0 = 1.-Omega_m0-Omega_r0
    E = np.sqrt(Omega_m0*np.exp(-3.*x) + Omega_r0*np.exp(-4.*x) + Omega_L0)
    return E
    
def comp_Omega_r(x, E, Omega_r0, Omega_m0, f_phi):
    Omega_r = Omega_r0*np.exp(-4.*x)/E/E
    return Omega_r

def comp_Omega_m(x, E, Omega_r0, Omega_m0, f_phi):
    Omega_m = Omega_m0*np.exp(-3.*x)/E/E
    return Omega_m

def comp_Omega_L(x, E, Omega_r0, Omega_m0, f_phi):
    Omega_phi0 = (1.-Omega_m0-Omega_r0)*f_phi
    Omega_L0 = 1.-Omega_m0-Omega_r0-Omega_phi0
    Omega_L = Omega_L0/E/E
    return Omega_L

def comp_E(x, phi_prime, Omega_r0, Omega_m0, f_phi):

    g_31 = comp_g_31(Omega_r0, Omega_m0, f_phi)
    k_1 = comp_k_1T(Omega_r0, Omega_m0, f_phi)
    Omega_phi0 = (1.-Omega_m0-Omega_r0)*f_phi
    Omega_L0 = 1.-Omega_m0-Omega_r0-Omega_phi0

    root_val = (k_1*phi_prime**2/6 - 1)**2 - 4*g_31*phi_prime**3*(Omega_L0 + Omega_m0*np.exp(-3.*x) + Omega_r0*np.exp(-4.*x))
    E = np.sqrt((1 - k_1*phi_prime**2/6 - np.sqrt(root_val))/2/g_31/phi_prime**3)
    return E
    
def comp_E_prime(x, E, phi_prime, Omega_r0, Omega_m0, f_phi):
    # phi'_0 = 1
    g_31 = comp_g_31(Omega_r0, Omega_m0, f_phi)
    k_1 = comp_k_1T(Omega_r0, Omega_m0, f_phi)

    Omega_r_val = comp_Omega_r(x, E, Omega_r0, Omega_m0, f_phi)
    Omega_L_val =comp_Omega_L(x, E, Omega_r0, Omega_m0, f_phi)
    
    term1 = (k_1*E*phi_prime*phi_prime/2 - E*(Omega_r_val - 3*Omega_L_val + 3)) * (k_1/phi_prime + 6*g_31*E*E)
    term2 = 9*g_31**2*E**5*phi_prime**3 + 3*g_31*k_1*E**3*phi_prime**2
    denominator = 2*(k_1 + 6*g_31*E*E) - 3*g_31*g_31*E**4*phi_prime**3
    
    return (term1+term2)/denominator


def comp_phi_primeprime(x, phi_prime, Omega_r0, Omega_m0, f_phi):
    g_31 = comp_g_31(Omega_r0, Omega_m0, f_phi)
    k_1 = comp_k_1T(Omega_r0, Omega_m0, f_phi)
    
    E = comp_E(x, phi_prime, Omega_r0, Omega_m0, f_phi)

    X_val = E**2*phi_prime**2/2
    E_prime_val = comp_E_prime(x, E, phi_prime, Omega_r0, Omega_m0, f_phi)
    
    nominator = 3*g_31*E*phi_prime*phi_prime*(E_prime_val + 3*E) + 3*k_1*phi_prime
    denominator = k_1 + 6*g_31*E*E*phi_prime
    return -1*nominator/denominator - E_prime_val*phi_prime/E


    
z_final = 0.
x_ini = np.log(1./(1.+30))
x_final = np.log(1./(1.+z_final))
x_arr = np.linspace(x_final, x_ini, 200)

Y0 = [1]

f_phi = 0.2

ans = scipy.integrate.solve_ivp(comp_phi_primeprime,[x_final,x_ini], Y0, t_eval=x_arr, method='LSODA', args=(cosmo_testing["Omega_g"], \
                cosmo_testing["Omega_m"], f_phi),atol=1e-7, rtol = 1e-7)
```

```python
# For today formalism 
def comp_k_1T(Omg_r, Omg_m, f_phi):
    return 6*f_phi*(Omg_m + Omg_r - 1)

def comp_g_31(Omega_r0, Omega_m0, f_phi):
    k_1 = comp_k_1T(Omega_r0, Omega_m0, f_phi)
    Omega_phi0 = (1.- Omega_m0-Omega_r0)*f_phi

    return Omega_phi0 - k_1/6

def comp_E_LCDM(x, Omega_r0, Omega_m0):
    Omega_L0 = 1.-Omega_m0-Omega_r0
    E = np.sqrt(Omega_m0*np.exp(-3.*x) + Omega_r0*np.exp(-4.*x) + Omega_L0)
    return E
    
def comp_Omega_r(x, E, Omega_r0, Omega_m0, f_phi):
    Omega_r = Omega_r0*np.exp(-4.*x)/E/E
    return Omega_r

def comp_Omega_m(x, E, Omega_r0, Omega_m0, f_phi):
    Omega_m = Omega_m0*np.exp(-3.*x)/E/E
    return Omega_m

def comp_Omega_L(x, E, Omega_r0, Omega_m0, f_phi):
    Omega_phi0 = (1.-Omega_m0-Omega_r0)*f_phi
    Omega_L0 = 1.-Omega_m0-Omega_r0-Omega_phi0
    Omega_L = Omega_L0/E/E
    return Omega_L

def comp_E(x, phi_prime, Omega_r0, Omega_m0, f_phi):

    g_31 = comp_g_31(Omega_r0, Omega_m0, f_phi)
    k_1 = comp_k_1T(Omega_r0, Omega_m0, f_phi)
    Omega_phi0 = (1.-Omega_m0-Omega_r0)*f_phi
    Omega_L0 = 1.-Omega_m0-Omega_r0-Omega_phi0

    root_val = (k_1*phi_prime**2/6 - 1)**2 - 4*g_31*phi_prime**3*(Omega_L0 + Omega_m0*np.exp(-3.*x) + Omega_r0*np.exp(-4.*x))
    E = np.sqrt((1 - k_1*phi_prime**2/6 - np.sqrt(root_val))/2/g_31/phi_prime**3)
    return E
    
def comp_E_prime(x, E, phi_prime, Omega_r0, Omega_m0, f_phi):
    # phi'_0 = 1
    g_31 = comp_g_31(Omega_r0, Omega_m0, f_phi)
    k_1 = comp_k_1T(Omega_r0, Omega_m0, f_phi)

    Omega_r_val = comp_Omega_r(x, E, Omega_r0, Omega_m0, f_phi)
    Omega_L_val =comp_Omega_L(x, E, Omega_r0, Omega_m0, f_phi)
    
    term1 = (k_1*E*phi_prime*phi_prime/2 - E*(Omega_r_val - 3*Omega_L_val + 3)) * (k_1/phi_prime + 6*g_31*E*E)
    term2 = 9*g_31**2*E**5*phi_prime**3 + 3*g_31*k_1*E**3*phi_prime**2
    denominator = 2*(k_1 + 6*g_31*E*E) - 3*g_31*g_31*E**4*phi_prime**3
    
    return (term1+term2)/denominator


def comp_phi_primeprime(x, phi_prime, Omega_r0, Omega_m0, f_phi):
    g_31 = comp_g_31(Omega_r0, Omega_m0, f_phi)
    k_1 = comp_k_1T(Omega_r0, Omega_m0, f_phi)
    
    E = comp_E(x, phi_prime, Omega_r0, Omega_m0, f_phi)

    X_val = E**2*phi_prime**2/2
    E_prime_val = comp_E_prime(x, E, phi_prime, Omega_r0, Omega_m0, f_phi)
    
    nominator = 3*g_31*E*phi_prime*phi_prime*(E_prime_val + 3*E) + 3*k_1*phi_prime
    denominator = k_1 + 6*g_31*E*E*phi_prime
    return -1*nominator/denominator - E_prime_val*phi_prime/E


    
z_final = 0.
x_ini = np.log(1./(1.+30))
x_final = np.log(1./(1.+z_final))
x_arr = np.linspace(x_final, x_ini, 200)

Y0 = [1]

f_phi = 0.2

ans = scipy.integrate.solve_ivp(comp_phi_primeprime,[x_final,x_ini], Y0, t_eval=x_arr, method='LSODA', args=(cosmo_testing["Omega_g"], \
                cosmo_testing["Omega_m"], f_phi),atol=1e-7, rtol = 1e-7)
```
